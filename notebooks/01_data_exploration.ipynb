## Selected Fields for Churn Analysis

The following columns were selected based on the Lending Club Data Dictionary and their relevance
to defining customer activity and churn behavior:

- **issue_d**: Month when the loan was funded.
- **last_pymnt_d**: Last recorded payment date, used as a proxy for customer activity.
- **loan_status**: Current status of the loan (e.g., Current, Fully Paid, Charged Off).
- **loan_amnt**: Original loan amount issued to the borrower.
- **total_pymnt**: Total payments received to date.
- **int_rate**: Interest rate assigned to the loan.
- **annual_inc**: Self-reported annual income provided by the borrower.
- **grade**: Lending Club assigned loan grade, representing credit risk category.
- **installment**: Monthly payment owed by the borrower.

These fields allow us to:

- Define churn as a 90-day period of inactivity.
- Analyze churn behavior relative to loan size and repayment history.
- Explore whether financial capacity (income), loan risk (grade, interest rate), and payment burden (installment) are associated with churn.

```python
# Select relevant columns for churn and risk analysis
cols = [
    "issue_d",
    "last_pymnt_d",
    "loan_status",
    "loan_amnt",
    "total_pymnt",
    "int_rate",
    "annual_inc",
    "grade",
    "installment"
]

df_churn = df[cols].copy()

# Parse dates
df_churn["issue_d"] = pd.to_datetime(df_churn["issue_d"], format="%b-%Y")
df_churn["last_pymnt_d"] = pd.to_datetime(df_churn["last_pymnt_d"], format="%b-%Y")

df_churn.head()
```

## Churn Definition & Feature Engineering

### 1. Churn Definition

In this analysis, customer churn is defined as a period of inactivity of 90 days or more since the borrower’s last recorded payment.

Unlike traditional financial churn definitions that rely on loan default or charge-off status, this approach focuses strictly on behavioral inactivity. This allows us to:

- Detect disengagement earlier.
- Avoid conflating churn with credit risk events.
- Maintain a behavior-driven definition independent of loan outcome.

#### Operational Definition

Days since last payment:

- days_since_last_payment = reference_date − last_pymnt_d

Churn classification rule:

- is_churned = True if days_since_last_payment ≥ 90
- is_churned = False otherwise

The reference date corresponds to the most recent date available in the dataset and represents the observation cutoff point.

---

### 2. Feature Engineering

To support churn analysis, several additional variables are derived from the selected fields.

---

#### 2.1 Days Since Last Payment

Derived from:

- last_pymnt_d

Purpose:

- Quantify customer inactivity.
- Serve as the primary behavioral indicator used to classify churn.

---

#### 2.2 Loan Age

loan_age_days = reference_date − issue_d

Purpose:

- Distinguish newly issued loans from mature loans.
- Prevent misclassification of recent loans as churned.
- Provide lifecycle context for repayment behavior.

This variable helps control for exposure time effects in churn analysis.

---
#### 2.3 Payment Performance Ratio

payment_ratio = total_pymnt / loan_amnt

Purpose:

- Measure repayment progress.
- Identify whether churn tends to occur early or late in the repayment lifecycle.

Interpretation:

- < 1 → Loan not fully repaid
- ≈ 1 → Loan nearly repaid
-         1 → Includes interest payments

---

#### 2.4 Financial Capacity Indicator

installment_to_income = installment / (annual_inc / 12)

Purpose:

- Approximate monthly payment burden.
- Evaluate whether financial strain is associated with higher churn likelihood.

Higher values suggest a greater proportion of income dedicated to loan repayment.

---

#### 2.5 Risk Indicators

The following existing variables capture borrower risk profile:

- grade
- int_rate

These variables reflect creditworthiness and may influence repayment continuity and churn probability.

---

### 3. Analytical Framework

The churn modeling framework separates borrower behavior into four dimensions:

- Behavioral signal → inactivity (days since last payment)
- Financial profile → income, installment, loan amount
- Risk profile → grade, interest rate
- Lifecycle stage → loan age, payment ratio

This structured approach enables a multidimensional understanding of churn behavior beyond simple loan default classification.


```python
reference_date = df_churn["last_pymnt_d"].max()
reference_date
```

```python
df_churn["days_since_last_payment"] = (
    reference_date - df_churn["last_pymnt_d"]
).dt.days

df_churn[["last_pymnt_d", "days_since_last_payment"]].head()
```

```python
CHURN_DAYS = 90

df_churn["is_churned"] = df_churn["days_since_last_payment"] > CHURN_DAYS

df_churn["is_churned"].value_counts(normalize=True)
```
